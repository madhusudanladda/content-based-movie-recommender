```python
import kagglehub
import pandas as pd
import os

# Download datasets
path1 = kagglehub.dataset_download("tmdb/tmdb-movie-metadata")
path2 = kagglehub.dataset_download("rounakbanik/the-movies-dataset")

print("TMDB Path:", path1)
print("Movies Dataset Path:", path2)
```

```python
# TMDB dataset
tmdb_movies = pd.read_csv(os.path.join(path1, "tmdb_5000_movies.csv"))
tmdb_credits = pd.read_csv(os.path.join(path1, "tmdb_5000_credits.csv"))

# Movies dataset
movies_metadata = pd.read_csv(os.path.join(path2, "movies_metadata.csv"), low_memory=False)
credits2 = pd.read_csv(os.path.join(path2, "credits.csv"))
keywords = pd.read_csv(os.path.join(path2, "keywords.csv"))
```

```python
# Convert IDs to numeric
movies_metadata['id'] = pd.to_numeric(movies_metadata['id'], errors='coerce')
keywords['id'] = pd.to_numeric(keywords['id'], errors='coerce')
credits2['id'] = pd.to_numeric(credits2['id'], errors='coerce')

# Drop null ids
movies_metadata.dropna(subset=['id'], inplace=True)
```

```python
movies_full = movies_metadata.merge(keywords, on='id')
movies_full = movies_full.merge(credits2, on='id')

movies_full.head()
```

```python
merged_movies = movies_full.merge(
    tmdb_movies,
    left_on='original_title',
    right_on='title',
    how='inner'
)

merged_movies.head()
```

```python
# Example — keep TMDB version
columns_to_drop = [col for col in merged_movies.columns if col.endswith('_y')]

merged_movies.drop(columns=columns_to_drop, inplace=True)

# Rename _x columns
merged_movies.columns = merged_movies.columns.str.replace('_x', '')
```

```python
merged_movies.shape
merged_movies.head()
```

```python
movies = merged_movies[[
    'title',
    'overview',
    'genres',
    'keywords',
    'cast',
    'crew',
    'vote_average'
]]
```

```python
import ast

def convert(text):
    try:
        return [i['name'] for i in ast.literal_eval(text)]
    except:
        return []

movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def get_cast(text):
    try:
        return [i['name'] for i in ast.literal_eval(text)[:3]]
    except:
        return []

movies['cast'] = movies['cast'].apply(get_cast)
```

```python
movies['overview'] = movies['overview'].fillna('')

movies['tags'] = (
    movies['overview'] + " " +
    movies['genres'].apply(lambda x: " ".join(x)) + " " +
    movies['keywords'].apply(lambda x: " ".join(x)) + " " +
    movies['cast'].apply(lambda x: " ".join(x))
)
```

```python
movies[['title','tags']].head()
```

```python
movies = movies[movies['tags'].notna()]
movies = movies.reset_index(drop=True)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words='english'
)

vectors = cv.fit_transform(movies['tags']).toarray()
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommend(movie):

    movie_index = movies[movies['title'] == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(movies.iloc[i[0]].title)
```

```python
recommend('Toy Story')
```

```python
def recommend(movie):

    movie = movie.lower()

    if movie not in movies['title'].str.lower().values:
        print("Movie not found!")
        return

    movie_index = movies[movies['title'].str.lower() == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    print("\nTop Recommendations:\n")

    for i in movies_list:
        print(movies.iloc[i[0]].title)
```

```python
import pickle

pickle.dump(movies, open('movies.pkl','wb'))
pickle.dump(similarity, open('similarity.pkl','wb'))
```
